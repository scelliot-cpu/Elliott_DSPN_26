## 1. Deriving the Maximum Likelihood Estimate for Simple Linear Regression (6 points)

Using the mean squared error (MSE) as your objective function (the thing you're trying to minimize when you fit your model) allows for a closed form solution to finding the maximum likelihood estimate (MLE) of your model parameters in linear regression. Let’s consider the simple, single predictor variable model, i.e. simple linear regression :  $Y= \beta_0 + \beta_1 X $.

a) Use algebra to show how you can expand out $MSE(\beta_0, \beta_1)$ to get from i to ii below.

> _i)_ $E[ (Y-(\beta_0 + \beta_1 X))^2]$

> _ii)_ $E[Y^2] -2 \beta_0E[Y]-2 \beta_1 Cov[X,Y]-2 \beta_1 E[X]E[Y]+ \beta_0^2 +2 \beta_0 \beta_1 E[X]+\beta_1^2 Var[X]+ \beta_1^2 (E[X])^2$






**Answer:**

E[(Y-(B0+B1X))^2]

E[(Y-B0-B1X)^2]

using (a-b-c)^2 = a^2 + b^2 + c^2 - 2ab + 2bc - 2ca:

E(Y^2 + B0^2 + B1X^2 - 2YB0 + 2B0B1X - 2YB1X)

use tip 2:
**E[Y^2]** + B0^2 + B1^2**E[X^2]** - 2B0**E[Y]** + 2B0B1**E[X]** - 2B1**E[YX]**

use tip 4: E[Y^2] + B0^2 + B1^2**Var(X)+E(X)^2** - 2B0E[Y] + 2B0B1E[X] - 2B1E[XY]

use tip 5: E[Y^2] + B0^2 + B1^2Var(X)+E(X)^2  - 2B0E[Y] + 2B0B1E[X] - 2B1**(Cov(X,Y)+E(X)E(Y)**

rewrite: E[Y^2] - 2B0E[Y] - 2B1(Cov(X,Y)+E(X)E(Y)) + B0^2 + 2B0B1E[X] + B1^2Var(X)+B1^2E(X)^2

b) Prove that the MLE of $\beta_0$ is $E[Y]- \beta_1 E[X]$ by taking the derivative of _ii_ above, with respect to $\beta_0$, setting the derivative to zero, and solving for $\beta_0$.

**Answer:**

Rule 5 from the tutorial says that since we are taking the derivative with respect to B0, we treat the other variables as constants, so all the variables that don't include B0, the derivative will be zero, leaving us with:

 -2B0[Y] + B0^2 + 2B0B1E[X]

d/dB0(-2B0[Y]) = -2E[Y]

d/dB0(B0^2) = 2B0

d/dB0(2B0B1E[X]) = 2B1E[X]

-2E[Y] + 2B0 + 2B1E[X] = 0

divide by 2: -E[Y] + B0 + B1E[X] = 0

B0 = E[Y] - B1E[X]

c) Prove that the MLE for $\beta_1$ is $Cov[X,Y]/Var[X]$ by taking the derivative of equation _ii_ above, with respect to $\beta_1$, setting the derivative to zero, and solving for $\beta_1$. *Hint: after you've simplified / expanded a bit, plug in the solution for $\beta_0$ from part b.*

**Answer:**

Same as before, variables that don't include B1, the derivate will be zero, leaving us with:

-2B1(Cov(X,Y)+E(X)E(Y)) + 2B0B1E[X] + B1^2Var(X)+ B1^2E(X)^2

d/dB1(-2B1(Cov(X,Y)) = -2Cov(X,Y)

d/dB1(-2B1(E(X)E(Y))) = -2E(X)E(Y)

d/dB1(2B0B1E(X)) = 2B0E(X)

d/dB1(B1^2Var(X)) = 2B1Var(X)

d/dB1(B1^2E(X)^2) = 2B1E(X)^2

0 = -2Cov(X,Y) -2E(X)E(Y) + 2B0E(X) + 2B1Var(X) + 2B1E(X)^2

0 = -Cov(X,Y) -E(X)E(Y) + B0E(X) + B1Var(X) + B1E(X)^2

Plug in derivative of B0:
0 = -Cov(X,Y) -E(X)E(Y) +  E[Y] - B1E[X](E(X)) + B1Var(X) + B1E(X)^2

0 = -Cov(X,Y) -E(X)E(Y) +  E[X]E[Y] - B1E[X]^2 + B1Var(X) + B1E(X)^2

Terms cancel out: 0 = -Cov(X,Y) + B1Var(X)

B1Var(X) = Cov(X,Y)

B1 = Cov(X,Y)/Var(X)
